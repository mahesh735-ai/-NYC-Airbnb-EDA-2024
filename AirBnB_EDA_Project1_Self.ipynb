# Final Insights: NYC Airbnb 2024 EDA Project

1. Data Cleaning & Quality Control
Data Volume: Dropped rows with missing values and removed 12 duplicate records, leaving a clean dataset of 20,736 listings.

Type Casting: Converted id and host_id into categorical object types to avoid accidental mathematical operations.

Outlier Capping: Removed realistic price distortions by capping the data to properties under $1,500 [ since   the  max price was skewed up to $100,000 while 75% of data was under $199 ].


2. Univariate Distribution Trends
Price Distribution: Nightly rates are heavily right-skewed, meaning the vast majority of listings belong to affordable, budget-friendly segments.

Availability Patterns: availability_365 shows a bimodal split—properties are either booked out completely (low availability) or run as full-time commercial listings (available 365 days).


3. Feature Engineering & Valuation
Metric Creation: Created a price_per_bed feature to measure the true structural value of listings across boroughs.

Borough Premium: Manhattan leads with the highest cost premium at $138.71 per bed, followed by Brooklyn ($99.79), while Staten Island is the cheapest ($67.72).


4. Bivariate Interactions (Key Relationships)
Room Type Hierarchy: Across all 5 boroughs, Entire homes/apartments consistently charge the highest prices, followed by Private rooms, Shared rooms, and Hotel rooms.

Review vs Price Paradox: The scatter plot proves that as reviews increase, prices drop. High customer traffic and bookings happen in affordable properties, not luxury ones.


5. Multivariate & Geographic Insights
Supply Dominance: The geographical map shows that Entire homes and Private rooms completely dominate NYC, while hotel and shared rooms are extremely rare.

Size Drives Price: The correlation matrix shows a solid positive link (0.42) between beds and price—capacity is the main driver of cost.

The Manhattan Effect: A negative correlation (-0.19) with longitude proves that moving east (away from Manhattan) naturally lowers property prices.

Calendar Factor: A near-zero correlation (0.05) between availability_365 and price shows that keeping a listing open all year doesn't let hosts charge higher rates.
